```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("simple_tabular_training_data.csv")
print("Dataset size:", df.shape)
print(df.head())
```

```python


grouped = (
    df.groupby(['City', 'Humidity_Level'])
      .size()
      .unstack(fill_value=0)
)

ax = grouped.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6),
    width=0.8,
    colormap='tab20'
)

# Replace long names with incremental IDs
ax.set_xticks(range(len(grouped)))
ax.set_xticklabels(range(1, len(grouped) + 1), rotation=0)

plt.xlabel('City')
plt.ylabel('count')
plt.title('Stacked Bar Chart')
plt.legend(title='Humidity Levels', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Group by one column
grouped = df.groupby('Humidity_Level').size()

# Create bar chart
ax = grouped.plot(
    kind='bar',
    figsize=(12, 6),
    width=0.8,
    color='steelblue'
)

# Replace long names with incremental IDs
ax.set_xticks(range(len(grouped)))
ax.set_xticklabels(range(1, len(grouped) + 1), rotation=0)

plt.xlabel('Humidity Level')
plt.ylabel('Count')
plt.title('Count by Humidity Level')
plt.tight_layout()
plt.show()
```
